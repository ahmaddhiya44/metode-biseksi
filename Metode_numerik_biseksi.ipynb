```python

import math

# 1. Fungsi persamaan nonlinear dari soal
def f(t):
    return 800 * math.exp(0.3 * t) - 2000 - 400 * t

# 2. Menentukan interval awal dan toleransi
a = 5.0
b = 6.0
toleransi = 0.0001
iterasi = 0

# 3. Memeriksa apakah interval memenuhi syarat Biseksi
if f(a) * f(b) >= 0:
    print("Interval awal tidak memenuhi syarat metode Biseksi.")
else:
    # 4. Menampilkan judul tabel iterasi
    print(
        f"{'Iterasi':>7} {'a':>10} {'b':>10} "
        f"{'c':>10} {'f(a)':>12} {'f(b)':>12} {'f(c)':>12}"
    )
    print("-" * 83)

    # 5. Melakukan iterasi sampai toleransi terpenuhi
    while True:
        iterasi += 1

        # Menghitung titik tengah interval
        c = (a + b) / 2

        # Menghitung nilai fungsi pada a, b, dan c
        fa = f(a)
        fb = f(b)
        fc = f(c)

        # Menampilkan hasil dengan 4 angka desimal
        print(
            f"{iterasi:7d} {a:10.4f} {b:10.4f} "
            f"{c:10.4f} {fa:12.4f} {fb:12.4f} {fc:12.4f}"
        )

        # Menghentikan iterasi jika toleransi terpenuhi
        if abs(fc) < toleransi or (b - a) / 2 < toleransi:
            break

        # Menentukan interval berikutnya
        if fa * fc < 0:
            b = c
        else:
            a = c

    # 6. Menampilkan hasil akhir
    print("\nHasil akhir:")
    print(f"Perkiraan titik impas t = {c:.4f} bulan")
    print(f"Nilai fungsi f(t) = {fc:.4f}")
```
